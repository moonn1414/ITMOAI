```python
df = pd.read_csv('HR-data.csv')
df.head()
```

Разбиваем дата сет на train и test



```python
x = df.drop(columns=['Attrition'])
y = df['Attrition']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=8)
```

2) Выберем более сложную ML-модель а именно RandomForestClassifier
---





```python
rf = RandomForestClassifier(random_state=8)
rf.fit(x_train, y_train)
y_preds = rf.predict(x_test)
rf.get_params()
```

Осуществим подбор гиперпараметров с scikit-learn методами на кросс-валидации а именно GridSearchCV, потому что он позволит перебрать возможные варианты наиболее ресурсоемким путем, с учетом использвования модели RandomForestClassifier

```python
parameters = {
    # Ограничим количество ветвей деревьев и установим шаг в 20, для ускорения работы модели и её возможности запуска в условиях малых мощностей
    'n_estimators': np.arange(40, 140, 20),
    'max_depth': list(range(3, 12)) + [None],
    'random_state': [8],
}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=8),
    param_grid=parameters,
    n_jobs=-1,
    verbose=1,
    cv=3,
)
```

```python
grid_search.fit(x_train, y_train)

best_parameters = grid_search.best_estimator_.get_params()
for param_name in sorted(parameters.keys()):
    print(f"{param_name}: {best_parameters[param_name]}")
```

Выпишем предложенные параметры модели RandomForestClassifier

`max_depth`: 10
`n_estimators`: 80
`random_state`: 8

```python
y_preds = rf.predict(x_test)
print(classification_report(y_test, y_preds))

#Для оценки модели в качестве выбранной метрики используем Accuracy
rf_accuracy = accuracy_score(y_test, y_preds)
print(f'Accuracy: {rf_accuracy:.4f}')
```

Произведем подбор гиперпараметров с optuna на кросс-валидации

```python
def chose(trial: optuna.Trial):
    max_depth = trial.suggest_int('max_depth', 2, 26, log=True)
    n_estimators = trial.suggest_int('n_estimators', 2, 100)
    max_features = trial.suggest_categorical('max_features', [None, 'sqrt', 'log2'])

    classifier_rf = RandomForestClassifier(
        max_depth=max_depth,
        n_estimators=n_estimators,
        max_features=max_features,
        random_state=8
    )

    # Кросс-валидация с возвратом средней точности
    score = cross_val_score(classifier_rf, x_train, y_train, cv=3, n_jobs=-1)
    return score.mean()  # Возвращаем среднюю точность кросс-валидации
```

```python
sampler = optuna.samplers.TPESampler(seed=8)
studyrf = optuna.create_study(direction="maximize", sampler=sampler)
studyrf.optimize(chose, n_trials=50)
```

Выведем параметры наилучшего подбора гиперпараметра

```python
print(studyrf.best_trial)
```

 Лучшие параметры

params= {`'max_depth': 18`, `'n_estimators': 35`, `'max_features': 'sqrt`'}

```python
studyrf.best_params
```

Результат с использованием optuna оказался хуже чем  в случае использования метода GridSearchCV

```python
Rforest = RandomForestClassifier(random_state=8,**studyrf.best_params)
Rforest.fit(x_train, y_train)

y_preds = Rforest.predict(x_test)
print(classification_report(y_test, y_preds))

#Для оценки модели в качестве выбранной метрики используем Accuracy
Rforest_accuracy = accuracy_score(y_test, y_preds)
print(f'Accuracy: {Rforest_accuracy:.4f}')
```

Построим графики с отображением признаков влияющих на результат работы модели и попробуем проинтерпретировать их


```python
forest_importances = pd.Series(rf.feature_importances_, index=x_train.columns)

fig, ax = plt.subplots()
std = np.std([tree.feature_importances_ for tree in rf.estimators_], axis=0)
forest_importances.plot.bar(yerr=std, ax=ax)
ax.set_title("Важные параметры")
ax.set_ylabel("Степень влияния",)
fig.tight_layout()
```

Используем Permutation importances и перемешаем данные по одному признаку и посмотрим на изменения, отобразим результат по убыванию влияния признака и выведем топ 7 факторов

```python
importances = permutation_importance(rf, x_test, y_test, n_repeats=10, random_state=8, n_jobs=2)
forest_importances = pd.Series(importances.importances_mean, index=x_train.columns)

# Сортируем и выбираем топ-7 самых важных признаков
top_forest_importances = forest_importances.sort_values(ascending=False).head(7)

# Получаем индексы топ-7 признаков как список
top_indices = list(top_forest_importances.index)

# Построение графика
fig, ax = plt.subplots(figsize=(10, 6))
top_forest_importances.plot.bar( yerr=[importances.importances_std[x_train.columns.get_loc(col)] for col in top_indices], ax=ax)
ax.set_title("Топ-7 признаков по степени влияния")
ax.set_ylabel("Степень влияния")
plt.xticks(rotation=45)
fig.tight_layout()
plt.show()
```
