```python
class Chocolate:
  def __init__(self, price:float, contnut:bool, qnut:int, add:str=None): # Инициализация объекта
    self.price = price
    self.add = add
    self.contnut = contnut
    self.qnut = qnut
    if self.price > 0: self.price = price
    else: raise ValueError('Цена не может быть 0')
  def show_my_chocolate(self):
    if self.add != None: print (f'Шоколадный батончик, добавка: {self.add}')
    else: print('Обычный шоколадный батончик')
  def count_nuts (self):
    print('*' * self.qnut)
  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'
```

```python
choco = Chocolate(50,True, 10,'Milk')
choco.show_my_chocolate()
choco1 = Chocolate(90,True, 6,'weed')
choco1.show_my_chocolate()
choco2 = Chocolate(20,False,5,)
choco2.show_my_chocolate()
```

```python
choco.count_nuts()
choco1.count_nuts()
choco2.count_nuts()
```
